# WHISTLER: GitHub Actions GPU Runner
This notebook turns your Google Colab instance (with a free NVIDIA T4 GPU) into a self-hosted runner for the WHISTLER repository.

### Instructions:
1. In the menu above, go to **Runtime** -> **Change runtime type** and select **T4 GPU**.
2. On GitHub, go to your repository **Settings** -> **Actions** -> **Runners**.
3. Click **New self-hosted runner** and select **Linux**.
4. Scroll down to the "Configure" section and find your token (it will look like `--token ABC123XYZ...`). Copy just the token string.
5. Paste your GitHub Username, Repository Name, and Token into the cell below, then run the cell!

```python
GITHUB_USER = "AmanBanik"
GITHUB_REPO = "WHISTLER"
RUNNER_TOKEN = "PASTE_YOUR_TOKEN_HERE"

import os
# GitHub Actions runners block root execution by default, but Colab runs as root.
os.environ['RUNNER_ALLOW_RUNASROOT'] = '1'

print("Downloading GitHub Actions Runner...")
!mkdir -p actions-runner && cd actions-runner \
    && curl -o actions-runner-linux-x64-2.319.1.tar.gz -L https://github.com/actions/runner/releases/download/v2.319.1/actions-runner-linux-x64-2.319.1.tar.gz \
    && tar xzf ./actions-runner-linux-x64-2.319.1.tar.gz

print("\nConfiguring Runner...")
!cd actions-runner && ./config.sh --url https://github.com/{GITHUB_USER}/{GITHUB_REPO} --token {RUNNER_TOKEN} --unattended --labels gpu --replace

print("\nStarting Runner... (This cell will run indefinitely, waiting to execute your CI jobs!)")
!cd actions-runner && ./run.sh
```
